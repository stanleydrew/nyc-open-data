### Import libraries:
* pandas: for analysis/creating the dataframe
* sodapy: for programmatic access to Socrata
* dotenv: library that finds the _.env_ file in your directory
* os: looks up API token

```python
import pandas as pd
from sodapy import Socrata
from dotenv import load_dotenv
import os
```

### Create query function

```python
def get_data(**data):
    '''Function for accessing NYC Open Data via the Socrata API.'''
    load_dotenv() # get token env variable
    client = Socrata(data["domain"], data["token"]) # use City of New York domain and dataset ID
    results = client.get_all(data["id"]) # use the get_all function to get all rows
    df = pd.DataFrame(results) # create dataframe from results
    return df
```

### Run query function
Plug in the different variables to access the data that you want: 
* domain: the domain of NYC Open Data (data.cityofnewyork.us)
* id: the ID of the dataset
* token: your Socrata API token. Store this in the .env file, never directly in your script

```python
df = get_data(
    domain = "data.cityofnewyork.us",
    id = "gdk4-mbsv",
    token=os.getenv("TOKEN")
)
```
